## 3. ndarray的创建

```python
# 基础的创建方法
arr = np.array([1,2,3], dtype=np.float64) # 方法
print(arr.ndim) # 属性
print(arr)
```

```python
# copy
arr1 = np.copy(arr) # 元素跟参数里的数组相同，但是不是一个数组了
print(arr1)
```

```python
# 预定义形状
# 全0数组
arr = np.zeros((2,3))   # (行，列)
print(arr)
# 全1数组
arr = np.ones((3,3))
print(arr)
# 未初始化的数组
arr = np.empty((4,3))
print(arr)
# 固定值数组
arr = np.full((3,4),10)
print(arr)
# 特殊方法 np.zeros_like()  复制参数的形状，其中的值全为0 还有ones_like full_like empty_like 都有
arr1 = np.zeros_like(arr)
print(arr1)
```

```python
# 等差数列 1 2 3 4 5
arr = np.arange(1,10,2) # np.arange(start, end, step) 左闭右开
print(arr)
```

```python
# 等间隔数列
arr = np.linspace(0,100,5,dtype=np.int64)   # start, end, space(间隔) 0  25  50  75  100
print(arr)
```

```python
# 对数间隔数列
arr = np.logspace(0,4,2,base=2) # start, end, 个数, base=数字(底数)默认为10
print(arr)
```

```python
# 特殊矩阵
# 单位矩阵：主对角线上的数字为1，其余数字为0
arr = np.eye(5,dtype=np.int64)
print(arr)
```

```python
# 对角矩阵：主对角线上为非0的数，其他的数字为0
arr = np.diag([1,2,3,4])
print(arr)
```

```python
# 随机数组的生成
# 生成0到1之间的随机浮点数（均匀分布）
arr = np.random.rand(3,4)   # (行，列)
print(arr)
```

```python
# 生成随机数列（正态分布）（-3到3之间）
# 两边小，中间大
arr = np.random.randn(3,4)
print(arr)
```

### 5.索引与切片

```python
# 一维数组的索引与切片
arr = np.random.randint(1,100,20)
print(arr)
print(arr[0])
# 获取全部的元素
print(arr[:])
# 获取指定索引之间的切片
print(arr[2:5]) # [left,right)
# slice 切片方法
print(arr[slice(2,5)])  # start, end, step
# 布尔索引：通过布尔条件来获取元素
print(arr[arr>10])
print(arr[(arr>10) & (arr<70)])
```

```python
# 二维数组的索引与切片
arr = np.random.randint(1,100,(4,8))
print(arr)
# 切片第1行的内容
print(arr[0])
print(arr[0,:])
print(arr[1,2:5])
print(arr[1,slice(2,5)])
# 布尔索引（在二维数组中使用布尔索引返回的是一维数组）
print(arr[arr>50])
# 布尔索引 对指定行进行布尔索引
print(arr[2][arr[2]>50])
```

### 5. ndarray的运算

```python
# 算数运算
a = np.array([1,2,3])
b = np.array([2,3,4])
print(a+b)
print(a-b)
print(a*b)
print(a ** 2)
```

```python
c = [1,2,3]
d = [4,5,6]
print(c+d)  # 列表相加是字符串拼接的概念
```

```python
# 数组和数组之间的算数运算
a = np.array([[1,3,4],[2,3,4],[4,5,6]])
b = np.array([[2,3,4],[6,7,8],[0,2,4]])
print(a+b)
print(a-b)
print(a*b)
print(a ** 2)
# 数组和标量之间的算数运算
print(a+2)
```

```python
# 广播机制 —— 重点
# 同一维度：相同、或者行或列有一个是1行/列
a = np.array([1,2,3])   # 1*3
'''a
1 2 3
1 2 3
1 2 3
'''
b = np.array([[4],[5],[6]]) # 3*1
'''b
4 4 4
5 5 5
6 6 6
'''
print(a + b)

# 错误案例
a = np.array([1,2])   # 1*2
b = np.array([[4,5,6]]) # 1*3
print(a+b)
```

```python
# 矩阵运算
a = np.array([[1,3,4],[2,3,4],[4,5,6]])
b = np.array([[2,3,4],[6,7,8],[0,2,4]])
print(a * b)    # 这个是数组相乘，对应位置相乘即可
print(a @ b)    # 这个是线性代数中矩阵的乘法运算
```

### 2. 统计函数

求和、计算平均值、计算中位数、标准差、方差

查找最大值、最小值

计算分位数、累积和、累积差

```python
arr = np.random.randint(1,20,8)
print(arr)
```

```python
# 求和
print(np.sum(arr))
```

```python
# 计算平均值    有两种mean 和 average，后者是加权平均值
print(np.mean(arr))
```

```python
# 计算中位数    从小到大排序，然后选取中间的数字  
# 偶数：中间两个数的平均值  奇数：排序后中间的数值
print(np.median(arr))
```

```python
# 计算标准差、方差      平均值相同，方差越小越稳定
print(np.var([1,2,100]))  # 方差
print(np.std([1,2,100]))  # 标准差
```

```python
# 计算最值  max和argmax     min和argmin
print(f"最大值：{np.max(arr)},最大值索引：{np.argmax(arr)}")
print(f"最小值：{np.min(arr)},最小值索引：{np.argmin(arr)}")
```

```python
# 分位数
# 中位数
print(np.median(arr))
# 分位数
np.random.seed(0)
arr = np.random.randint(1,20,4)
print(arr)
print(np.percentile(arr,25))
print(np.percentile(arr,80))
"""
------------------
1   4    13     16

首先计算一共有多少个区间(n - 1)     4 - 1 = 3
计算的是80%分位数 = 0.8 * 3 = 2.4
可以看到是在2 ~ 3 这个区间内 也就是 13 ~ 16之间
然后用(16 - 13) * 0.4 = 1.2
这里的0.4 是2.4的小数部分
然后再用 13 + 1.2 = 14.2
就能得到 80% 中位数了
"""
```

```python
# 累积和、累积积
arr = np.array([1,3,4,5,7])
print(np.sum(arr))
print(np.cumsum(arr))   # 前n项的和，例如arr[1]是第0项和第1项的和
print(np.cumprod(arr))  # 前n项的积
```

比较函数：
    
    比较是否大于、小于、等于
    
    逻辑与、或、非
    
    检查数组中是否有一个True、是否所有的都为True、自定义条件

```python
# 是否大于
print(np.greater([1,2,3,4,5],2))
# 是否小于
print(np.less([1,2,3,4,5],3))
# 是否等于
print(np.equal([1,3,4,5],3))
print(np.equal([1,3,4,5],[1,2,3,5]))    # 两个数组进行比较，按相应索引进行对比
```

```python
# 逻辑与
print(np.logical_and([1,0],[1,1]))  # 逐元素逻辑与（对应位置比较）这里容易混淆，第一个元素和第一个元素进行逻辑运算
# 逻辑非
print(np.logical_not([1,0]))
# 逻辑或
print(np.logical_or([0,0],[0,1]))
```

```python
# 检查元素是否至少有一个元素为True
print(np.any([0,0,0,0,0]))
# 检查元素是否全部元素为True
print(np.all([0,1,1,0,0]))
```

```python
# 自定义条件
# print(np.where(条件,符合条件的,不符合条件的))
arr = np.array([1,2,3,4,5,6])
print(np.where(arr>3,arr,0))
```

```python
score = np.random.randint(50,101,20)
print(score)
print(np.where(score>60,"及格","不及格"))
```

```python
print(np.where(
    score<60,"不及格",np.where(
        score<80,"良好","优秀"
    )
))
```

```python
 # np.select(条件,返回的结果,default="默认值")
 print(np.select([score<60,(score<=80)&(score>=60),score>80],["不及格","良好","优秀"],default="未知"))  # 实测缺一不可，不然就会报错
```

```python
"""
题目1：温度数据分析
某城市一周的最高气温℃为[28, 30, 29, 31, 32, 30, 29]。
- 计算平均气温，最高气温和最低气温
- 找出气温超过30℃的天数
"""
arr = np.array([28, 30, 29, 31, 32, 30, 29])
print(f"平均气温是：{np.round(np.mean(arr),1)}，最高气温是：{np.max(arr)}，最低气温是：{np.min(arr)}")
print(f"气温超过30℃的天数为：{np.size(arr[arr>30])}")
```

```python
"""
题目 2：学生成绩统计
某班级 5 名学生的数学成绩为 [85,90,78,92,88]。
- 计算成绩的平均分、中位数和标准差。
- 将成绩转换为十分制（假设满分为 10）。
"""
arr = np.array([58,90,78,92,88])
print(f"平均分为：{np.mean(arr)}，中位数为：{np.median(arr)}，标准差为：{np.std(arr)}")
print(arr / 10)
```

```python
"""
题目 3：矩阵运算
给定矩阵 A = [[1, 2], [3, 4]] 和 B = [[5, 6], [7, 8]]。
- 计算 A + B 和 A * B（逐元素乘法）。
- 计算 A 和 B 的矩阵乘法（点积）。
"""
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(f"A + B = \n{ A + B }\nA * B = \n{A * B}\n矩阵A与矩阵B相乘 = \n{A @ B}")
```

```python
"""
题目 4：随机数据生成

生成一个（3，4）的随机整数数组，范围 [0, 10)。

计算每列的最大值和每行的最小值。

将数组中的所有奇数替换为 -1。
"""
np.random.seed(0)
arr = np.random.randint(0,10,[3,4])
print(arr)
print(f"每列的最大值：{np.max(arr,axis=0)}")    # axis参数表示轴，0表示列，1表示行
print(f"每行的最小值：{np.min(arr,axis=1)}")    
print(np.where(arr%2==1,-1,arr))
```

```python
"""
题目 5：数组变形

创建一个 1 到 12 的一维数组，并转换为 (3, 4) 的二维数组。

计算每行的和与每列的平均值。

将数组展平为一维数组。
"""
arr = np.arange(1,13,1)
print(arr)
new_arr = np.reshape(arr,[3,4])
print(new_arr)
print(f"每行的和：{np.sum(new_arr,axis=1)}")
print(f"每列的平均值：{np.mean(new_arr,axis=0)}")
print(new_arr.reshape(12))
```

```python
"""
生成一个（5，5）的随机数组，范围[0, 20]。

找出数组中大于 10 的元素。

将所有大于 10 的元素替换为 0。
"""
np.random.seed(0)
arr = np.random.randint(0,20,[5,5])
print(arr)
print(arr[arr>10])
print(np.where(arr>10,0,arr))
```

```python
"""
题目 7：统计函数应用

某公司 6 个月的销售额（万元）为 [120, 135, 110, 125, 130, 140]。

计算销售额的总和、均值和方差。

找出销售额最高的月份和最低的月份。
"""
arr = np.array([120, 135, 110, 125, 130, 140])
print(f"销售额的总和：{np.sum(arr)}，均值：{np.mean(arr)}，方差：{arr.var()}")
print(f"销售额最高的月份：{np.argmax(arr)+1}，最低的月份：{arr.argmin()+1}")
```

```python
"""
题目 8：数组拼接

给定 A = [1, 2, 3] 和 B = [4, 5, 6]。将 A 和 B 水平拼接为一个新数组。将 A 和 B 垂直拼接为一个新数组。
"""
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])
C = np.concatenate((A,B))   # 水平拼接
print(C)
print(C.reshape(2,3))
```

```python
"""
题目 9：唯一值与排序

给定数组 [2, 1, 2, 3, 1, 4, 3]。

找出数组中的唯一值并排序。

计算每个唯一值出现的次数。
"""
arr = np.array([2, 1, 2, 3, 1, 4, 3])
unique,counts = np.unique(arr,return_counts=True)
print(unique,"\n",counts)
```

```python
"""
题目：综合应用

某商店 5 天的销售额（万元）和成本（万元）如下：
销售额：[20, 25, 22, 30, 28]
成本：[15, 18, 16, 22, 20]

计算每天的利润（销售额 - 成本）。

计算利润的平均值和标准差。

找出利润最高的天数。
"""
smoney = np.array([20, 25, 22, 30, 28])
money = np.array([15, 18, 16, 22, 20])
lirun = smoney - money
print(f"每天的利润：{lirun}，利润的平均值：{lirun.mean()}，标准差：{lirun.std()}")
print(f"利润最高的天数为：{len(lirun[lirun==np.max(lirun)])}")

```
