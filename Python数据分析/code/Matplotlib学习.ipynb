```python
# 绘制折线图
import matplotlib.pyplot as plt

# 配置中文
from matplotlib import rcParams # 字体
rcParams['font.family'] = 'SimHei'  # windows是SimHei，mac是AppleGothic
rcParams['font.size'] = 12

# 创建图表，设置大小
# figure()函数：创建一个图表，参数有：figsize、dpi、facecolor、edgecolor、num、clear、frameon
# figsize：图表的大小，单位是英寸
# dpi：图表的像素密度
# facecolor：图表的背景颜色
# edgecolor：图表的边框颜色
# num：图表的编号
# clear：是否清除之前的图表
# frameon：是否显示图表的边框
plt.figure(figsize=(10,5))  # 画布的概念

# 要绘图的数据
month = ['1月','2月','3月','4月','5月','6月','7月','8月','9月','10月','11月','12月']
sales = [100,124,234,450,400,350,400,450,500,550,600,650]

# 再添加一组数据
sales2 = [120,140,160,180,200,220,240,260,280,300,320,340]

# 绘制折线图
plt.plot(month,sales,label='产品A',color='blue',linewidth=2)
plt.plot(month,sales2,label='产品B',color='red',marker='o',linestyle='--')

# 添加标题
plt.title('2025年销售量趋势',color='red',fontsize=20)
# 添加坐标轴标签
plt.xlabel('月份',fontsize=16,color='purple')
plt.ylabel('销售量（万元）',fontsize=16,color='purple')
# 添加图例
plt.legend(loc='upper left')
# 背景添加网格线
# grid()函数：添加网格线，参数有：alpha、axis,linestyle,color
# alpha：网格线的透明度，范围是0到1
# linestyle：网格线的样式，默认是实线，'-'代表实线，':'代表点线，'--'代表虚线，'.'代表点线
# axis：指定在哪个轴上添加网格线，默认是X和Y轴都有
# 例如：plt.grid(alpha=0.5, axis='x')
# 例如：plt.grid(alpha=0.5)
plt.grid(alpha=0.5,linestyle='--')

# 设置刻度字体大小
# rotation：刻度标签的旋转角度，默认是0度
# fontsize：刻度标签的字体大小，默认是10
plt.xticks(fontsize=12,rotation=45)
plt.yticks(fontsize=12)

# 设置y轴的范围
plt.ylim(0,700)

# 在每个数据点上显示数值
for x, y in zip(month,sales):
    # ha（horizontalalignment）：水平对齐，默认是center，参数有：left、center、right
    # va（verticalalignment）：垂直对齐，默认是center，参数有：top、bottom、center
    plt.text(x,y,str(y),ha='center',va='bottom')
#这里可以封装成函数，方便以后使用：
def add_label(x_list,y_list):
    for x,y in zip(x_list,y_list):
        plt.text(x,y,str(y),ha='center',va='bottom')
add_label(month,sales2) #直接给产品B也加上标签

# 显示图表
plt.show()
```

```python
# 创建图表，设置大小
plt.figure(figsize=(10,5))

# 要绘图的数据
subjects = ['语文','数学','英语','物理','化学','生物']
scores = [80,90,75,95,85,92]

# 绘制柱状图
plt.bar(subjects,scores,label='成绩',width=0.5,color='orange')

# 在每个数据点上显示数值
add_label(subjects,scores)

# 显示图例
plt.legend(loc='upper left')

# 添加标题
plt.title('期中考试平均成绩',color='red',fontsize=20)

# 添加坐标轴标签
plt.xlabel('科目',fontsize=16,color='purple')
plt.ylabel('成绩',fontsize=16,color='purple')

# 添加坐标轴范围
plt.ylim(0,100)

# 显示网格
plt.grid(axis='y',linestyle='--',alpha=0.5)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 创建图表，设置大小
plt.figure(figsize=(10,5))

# 要绘图的数据
things = ['学习','工作','睡觉','吃饭','其他']
time = [6,4,8,1,5]
colors=['#FFC0CB','#ADD8E6','#DDA0DD','#98FB98','#FFF6B7']

# 绘制环形图
# autopct 自动添加百分比标签,%.1f 表示保留一位小数,%% 表示添加百分号符号
# startangle 起始角度,默认是0度
plt.pie(time,labels=things,
        autopct='%.1f%%',
        startangle=90,
        colors=colors,
        wedgeprops={'width':0.6},   # 设置圆环的宽度百分比
        pctdistance=0.6 # 设置百分比数字的位置
        )

# 添加标题
plt.title('一天的时间分布',color='blue',fontsize=20)
plt.text(0,0,'总计24小时',ha='center',va='center',fontsize=16)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 创建图表，设置大小
plt.figure(figsize=(10,5))

# 要绘图的数据
things = ['学习','工作','睡觉','吃饭','其他']
time = [6,4,8,1,5]
colors=['#FFC0CB','#ADD8E6','#DDA0DD','#98FB98','#FFF6B7']
explode=[0.1,0,0,0,0]   # 设置突出块的位置

# 绘制爆炸式饼图
# autopct 自动添加百分比标签,%.1f 表示保留一位小数,%% 表示添加百分号符号
# startangle 起始角度,默认是0度
# explode 设置突出块的位置
plt.pie(time,labels=things,autopct='%.1f%%',startangle=90,colors=colors,explode=explode)

# 添加标题
plt.title('一天的时间分布',color='blue',fontsize=20)

# 自动优化排版
plt.tight_layout()

# 显示图表
plt.show()
```

```python
# 绘制散点图
plt.figure(figsize=(10,5))

# 要绘图的数据
scores = [50,55,60,65,70,75,80]
hours = [1,2,3,4,5,6,7]

# 绘制散点图
# 和折线图的区别是散点图没有连接线,折线图有连接线
plt.scatter(hours,scores)

# 显示图表
plt.show()
```

```python
# 绘制散点图
plt.figure(figsize=(10,8))

# 要绘图的数据
# 生成随机数据
import random
x = []
y = []
for i in range(200):
    tmp = random.uniform(0,10)
    x.append(tmp)
    tmp2 = 2*tmp+random.gauss(0,2)
    y.append(tmp2)
# 绘制散点图
# 和折线图的区别是散点图没有连接线,折线图有连接线
plt.scatter(x,y,
            color='blue',
            alpha=0.5,
            s=40,    # 设置圆点的大小
            label='数据'
            )

# 添加标题
plt.title('X变量与Y变量的关系图')

# 添加图例
plt.legend(loc='upper left')

# 设置坐标轴的标签
plt.xlabel('X自变量')
plt.ylabel('Y因变量')

# 设置网格线
plt.grid(True,linestyle='--',alpha=0.5)

# 画一个回归线
plt.plot([0,10],[0,20],color='red',linewidth=2)

# 显示图表
plt.show()
```

```python
# 绘制箱线图
plt.figure(figsize=(8,6))

# 模拟三门课的成绩
data = {
    '语文':[80,90,85,77,88,92,88,90,85,92],
    '数学':[85,95,88,92,90,85,92,88,90,92],
    '英语':[90,85,92,88,90,85,92,88,90,92]
}

# 绘制箱线图
plt.boxplot(data.values(),tick_labels=data.keys())

# 设置标题
plt.title('各科成绩分布（箱线图）')

# 显示分隔线
plt.grid(True,axis='y',linestyle='--',alpha=0.5)

# 显示图表
plt.show()
```

```python
# 多个图的绘制方法
# 要绘图的数据
import matplotlib.pyplot as plt
month = ['1月','2月','3月','4月','5月','6月','7月','8月','9月','10月','11月','12月']
sales = [100,124,234,450,400,350,400,450,500,550,600,650]
f1 = plt.subplot(2,2,1)  # 2行2列，第1个子图
f1.plot(month,sales)
f1.tick_params(axis='x',rotation=45)
f2 = plt.subplot(2,2,2)  # 2行2列，第2个子图
f2.bar(month,sales)
f2.tick_params(axis='x',rotation=45)
f3 = plt.subplot(2,2,3)  # 2行2列，第3个子图
f3.scatter(month,sales)
f3.tick_params(axis='x',rotation=45)
f4 = plt.subplot(2,2,4)  # 2行2列，第4个子图
f4.barh(month,sales)
plt.show()
```

```python
# 分析案例：温度分析
# 1. 导入库
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.sans-serif'] = ['SimHei']

# 2. 导入数据
df = pd.read_csv('data/weather.csv')
df.head()

# 3.1 绘制气温的趋势变化图
df['date'] = pd.to_datetime(df['date'])
df = df[df['date'].dt.year==2015]
plt.figure(figsize=(10,10))
plt.plot(df['date'], df['temp_max'], label='最高气温')
plt.plot(df['date'], df['temp_min'], label='最低气温')
plt.legend(loc='upper left')
plt.title('2015年气温趋势变化图')
plt.xlabel('日期')
plt.ylabel('气温')
plt.show()



# 3.2 降水量的直方图
plt.hist(df['precipitation'], bins=5)
plt.title('2015年降水量直方图')
plt.xlabel('降水量')
plt.ylabel('次数')
plt.show()
```

```python
# 使用seaborn绘图
import seaborn as sns
penguins = pd.read_csv('data/penguins.csv')
penguins.dropna(inplace=True)
penguins.info()
# 直方图
sns.histplot(data=penguins, x='species')
```

```python
# 核密度估计图：
# 绘制喙长度的核密度估计图
sns.kdeplot(data=penguins, x='bill_length_mm')
```

```python
# 绘制一个直方图，并且添加核密度估计曲线
sns.histplot(data=penguins,x='bill_length_mm',kde=True)
```

```python
# 计数图
# 绘制不同岛屿企鹅数量的计数图
sns.countplot(data=penguins,x='island')
```

```python
sns.histplot(data=penguins,x='island')
# 将直方图和计数图进行比较，可以看出相同数据下，直方图的柱状图高度和计数图的柱状图高度是相同的
```

```python
# 散点图
# 绘制横轴为体重，纵轴为脚蹼长度的散点图。可通过hue参数设置不同组别进行比较
sns.scatterplot(data=penguins,x='body_mass_g',y='flipper_length_mm',hue='sex')
```

```python
# 蜂窝图
# 通过jointplot()函数，设置kind="hex"来绘制蜂窝图
sns.jointplot(data=penguins,x='body_mass_g',y='flipper_length_mm',kind='hex')
```

```python
# 二维核密度图
# 通过kdeplot()函数，同时设置x参数和y参数来绘制二维核密度估计图。
sns.kdeplot(data=penguins,x='body_mass_g',y='flipper_length_mm')
```

```python
# 条形图
# data：数据集
# x：横轴变量
# y：纵轴变量
# estimator：统计函数，默认是mean
# errorbar：是否显示误差条，默认是True
sns.barplot(data=penguins,x='species',y='bill_length_mm',estimator='mean',errorbar=None)
```

```python
# 成对关系图
# data：数据集
# hue：分组变量，默认是None
sns.pairplot(data=penguins,hue='species')
```
