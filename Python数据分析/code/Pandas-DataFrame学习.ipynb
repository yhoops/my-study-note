## DataFrame 创建方式

```python
# dataframe的创建方式
import pandas as pd
import numpy as np
# 通过Series来创建
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({"第1列":s1,"第2列":s2})
print(type(df))
print(type(df["第1列"]))
# 通过字典来创建
'''
dataframe的每一列都是一个Series对象，字典的key就是列名，value就是数据
columns参数可以指定列的顺序，index参数可以指定行索引
'''
df = pd.DataFrame({
    "姓名":["张三","李四","王五"],
    "年龄":[18,19,20],
    "性别":["男","女","男"],
    "成绩":[90,95,85]
},index=[1,2,3],columns=["姓名","年龄","成绩","性别"]
)
df
```

```python
# dataframe的属性
print('行索引：',df.index)
print('列标签：',df.columns)
print('值：',df.values)
print('维度：',df.ndim)
print('形状：',df.shape)
print('元素个数：',df.size)
print('\n数据类型：\n',df.dtypes)
```

```python
print('行列转置：\n',df.T)
print(df.T.index)
print(df.T.columns)
```

```python
# 获取元素 loc iloc at iat
# 获取某行数据
print(df)
print(df.loc[1])    # 显式索引
print(df.iloc[0])   # 隐式索引
```

```python
# 获取某列数据
print(df)
print(df.loc[:,"姓名"])    # 显式索引
print(df.iloc[:,0])       # 隐式索引
print(df["姓名"])          # 直接通过列名获取   最常用
print(df.姓名)            # 直接通过属性获取
```

```python
# 单个元素，at和loc获取单个元素其实差不多，只是at的速度更快一些
print(df.at[1,"姓名"])    # 显式索引，应该获取到张三
print(df.iat[0,0])       # 隐式索引，应该获取到张三
print(df.loc[1,"姓名"])    # 显式索引，应该获取到张三
print(df.iloc[0,0])       # 隐式索引，应该获取到张三
```

```python
# 获取单列数据
print(df["成绩"])
print(df.成绩)
```

```python
# 查看部分数据
print(df.head(2))  # 查看前两行数据
print(df.tail(2))  # 查看后两行数据
```

```python
# 布尔索引
# 找出score > 85的
print(df[df.成绩 > 85])
# 找出score > 85 并且 年龄<19的
print(df[(df.成绩 > 85) & (df.年龄 < 19)])
```

```python
'''
案例 1：学生成绩分析
场景：某班级的学生成绩数据如下，请完成以下任务：
1．计算每位学生的总分和平均分。
2．找出数学成绩高于 90 分或英语成绩高于 85 分的学生。
3．按总分从高到低排序，并输出前 3 名学生。
'''
import pandas as pd
data = {
    '姓名': ['张三', '李四', '王五', '赵六', '钱七'],
    '数学': [85, 92, 78, 88, 95],
    '英语': [90, 88, 85, 92, 80],
    '物理': [75, 80, 88, 85, 90]
}
```

```python
scores = pd.DataFrame(data)
# 1．计算每位学生的总分和平均分。
scores['总分'] = scores[['数学', '英语', '物理']].sum(axis=1)
scores['平均分'] = scores[['数学', '英语', '物理']].mean(axis=1).round(1)  # 保留1位小数
# 2．找出数学成绩高于 90 分或英语成绩高于 85 分的学生。
print(scores[(scores['数学'] > 90) | (scores['英语'] > 85)])
# 3．按总分从高到低排序，并输出前 3 名学生。
print(scores.sort_values(by='总分', ascending=False).head(3))
```

```python
'''
案例 2：销售数据分析
场景：某公司销售数据如下，请完成以下任务：
1．计算每种产品的总销售额（销售额 = 单价 × 销量）。
2．找出销售额最高的产品。
3．按销售额从高到低排序，并输出所有产品信息。
'''
import pandas as pd
data = {
    '产品名称': ['A', 'B', 'C', 'D'],
    '单价': [100, 150, 200, 120],
    '销量': [50, 30, 20, 40]
}
df = pd.DataFrame(data)
```

```python
# 1．计算每种产品的总销售额（销售额 = 单价 × 销量）。
df['销售额'] = df['单价'] * df['销量']
df
# 2．找出销售额最高的产品。
print(df.nlargest(1, columns=['销售额']))
# 3．按销售额从高到低排序，并输出所有产品信息。
df.sort_values(by='销售额', ascending=False)
```

```python
'''
案例 3：电商用户行为分析
场景：某电商平台的用户行为数据如下，请完成以下任务：
1．计算每位用户的总消费金额（消费金额 = 商品单价 × 购买数量）
2．找出消费金额最高的用户，并输出其所有信息
3．计算所有用户的平均消费金额（保留 2 位小数）
4．统计电子产品的总购买数量
'''
import pandas as pd

data = {
    '用户ID': [101, 102, 103, 104, 105],
    '用户名': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    '商品类别': ['电子产品', '服饰', '电子产品', '家居', '服饰'],
    '商品单价': [1200, 300, 800, 150, 200],
    '购买数量': [1, 3, 2, 5, 4]
}
df = pd.DataFrame (data)
df
```

```python
# 1．计算每位用户的总消费金额（消费金额 = 商品单价 × 购买数量）
df['总消费金额'] = df['商品单价'] * df['购买数量']
df
# 2．找出消费金额最高的用户，并输出其所有信息
print(df.nlargest(1, columns=['总消费金额']))
# 3．计算所有用户的平均消费金额（保留 2 位小数）
print(df['总消费金额'].mean().round(2))
# 4．统计电子产品的总购买数量
print(df[df['商品类别'] == '电子产品']['购买数量'].sum())
```
