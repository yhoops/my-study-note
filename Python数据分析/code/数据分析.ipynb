```python
# 数据处理
# 缺失值的处理
import pandas as pd
import numpy as np
s = pd.Series([1,2,3,np.nan,5,None,pd.NA])  # 这三种都是缺失值
df = pd.DataFrame([[1,2,3],[4,np.nan,6],[7,None,pd.NA]], columns=['第一列','第二列','第三列'])  # 这三种都是缺失值
print(s)
print(df)
# 查看是否是缺失值
print(s.isna())  # True表示缺失值
print(s.isnull())  # True表示缺失值
print(df.isna())
print(df.isnull())
print(s.isna().sum())  # 缺失值的数量
print(df.isna().sum())  # 缺失值的数量
```

```python
# 剔除缺失值
print(s.dropna())
print(df)
print(df.dropna())  # 一行内有缺失值，整行会被剔除
print(df.dropna(how='all'))  # 一行内全是缺失值，整行会被剔除
print(df.dropna(thresh=2))  # 至少有2个非缺失值，整行才会保留
print(df.dropna(axis=1))  # 一列内有缺失值，整列会被剔除
print(df.dropna(subset=['第二列']))  # 如果某列有缺失值，则那一行会被剔除
```

```python
# 数据分箱  pd.cut(x,bins,labels)  # 分箱，返回分箱的区间
# x: 要分箱的列
# bins: 分箱的区间
# labels: 分箱的标签
import pandas as pd
df = pd.read_csv('data/employees.csv',encoding='utf-8')
df.head(10)
```

```python
df1 = df.head(10)[['employee_id','salary']]
df1
```

```python
pd.cut(df1.salary,bins=2)   # bins=n，分成n段区间，起始值和结束值是所有数据的最小值和最大值
# 4180.2 - 14100.0 - 24000.0
pd.cut(df1.salary,bins=3).value_counts()  # 查看每个区间的数量
pd.cut(df1.salary,bins=[0,10000,20000,30000]).value_counts()   # bins=list，指定分箱的区间
# 0 - 10000 - 20000 - 30000
df1['收入范围'] = pd.cut(df1.salary,bins=[0,10000,20000,30000],labels=['低','中','高'])    # labels=list，指定每个区间的标签
pd.qcut(df1.salary,3).value_counts()   # qcut: 分位数分箱，q=n，分成n段区间
```

```python
# 分组聚合
# df.groupby('分组的字段')['聚合的字段'].聚合函数()
import pandas as pd
df = pd.read_csv('data/employees.csv')
print(df['department_id'].isna().sum())
df.dropna(subset=['department_id'],inplace=True)
print(df['department_id'].isna().sum())
df['department_id'] = df['department_id'].astype('int16')
# 计算不同部门的平均薪资
print(df.groupby('department_id').groups)   # 显示每个部门的索引，查看分组
print(df.groupby('department_id').get_group(10))    # 查看具体某个分组数据
df2 =df.groupby('department_id')['salary'].mean().round(2)
df2 = df2.reset_index()
df2.sort_values('salary',ascending=False)
```

```python
# 计算不同部门不同岗位的人的平均薪资
df3 = df.groupby(['department_id','job_id'])['salary'].mean().round(2)  # 单层返回的是Series
df3 = df.groupby(['department_id','job_id'])[['salary']].mean().round(2)    # 双层返回的是DataFrame，列名是salary
df3 = df3.reset_index()
df3 = df3.sort_values('salary',ascending=False)
df3
```

```python
# 企鹅数据分析
# 1. 导入必要库
import pandas as pd
# 2. 导入数据
df = pd.read_csv('data/penguins.csv')
df.head(5)
df.info()
```

```python
# 3. 数据清洗
# 3.1 缺失值处理
print(df.isna().sum())
df.dropna(inplace=True)
```

```python
# 4. 数据特征构造
df['sex'] = df['sex'].astype('category')
df['bill_ratio'] = df['bill_length_mm'] / df['bill_depth_mm']
```

```python
# 5. 数据分析
# 5.1 数据分箱-把体重分为低中高三个等级
labels = ['低','中','高']
df['mass_level'] = pd.cut(df['body_mass_g'],bins=3,labels=labels)
print(df['mass_level'].value_counts())
# 按岛屿，性别分组分析
# 只能算一种指标
# df.groupby('sex')['body_mass_g'].mean()
# agg可以一次性算多个指标：均值、数量、最大最小、标准差
# df.groupby('sex')['body_mass_g'].agg(['mean','count','max','min'])
df.groupby(['sex','island']).agg(
    {
        'body_mass_g':['mean','count']
    }
)
```

```python
# 1. 导入库
import pandas as pd
import numpy as np
# 2. 导入数据
df = pd.read_csv('data/sleep.csv')
df.info()
df.describe()
```

```python
# 3. 数据清洗
df.isna().sum()
df.drop(columns='sleep_disorder',inplace=True)
```

```python
# 4. 数据特征构造
df['gender'] = df['gender'].astype('category')
df['occupation'] = df['occupation'].astype('category')
df['bmi_category'] = df['bmi_category'].astype('category')
df[['high','low']] = df['blood_pressure'].str.split('/',expand=True)
# 睡眠质量的分箱
labels = ['差','中','优']
df['sleep_quality_level'] = pd.cut(df['sleep_quality'],bins=3,labels=labels)
age_labels = ['青年','中年','老年']
df['age_level'] = pd.cut(df['age'],bins=[0,30,60,120],labels=age_labels)
```

```python
# 5. 数据的统计与分析
print(df['bmi_category'].value_counts())
```

```python
# 根据不同的bmi分组，睡眠质量
df.groupby(['age_level','bmi_category']).agg(
    {
        'sleep_duration':'mean',
        'sleep_quality':'mean',
        'stress_level':'mean'
    }
)
```
