# Series 的创建

```python
# Series的创建
import pandas as pd
from pygments.lexers.sql import sqlite_prompt_re
from tornado.http1connection import parse_int

s = pd.Series([1,2,3,4,5])
print(s)
# 自定义索引
s = pd.Series([1,2,3,4,5],index=['A','B','C','D','E'])
print(s)
# 定义标签 name
s = pd.Series([1,2,3,4,5],index=['A','B','C','D','E'], name='月份')
print(s)
```

```python
# 通过字典来创建
s = pd.Series({"a":1,"b":2,"c":3,"d":4,"e":5})
print(s)
# 通过已有的Series来创建
s1 = pd.Series(s,index=["a","c"])
print(s1)
```

```python
# series 的属性
print(s.index)
print(s.values)
print(s.shape,s.ndim,s.size)
s.name = "test"
print(s.dtype,s.name)
print(s.loc['a'])  # 显式索引获取元素
print(s.iloc[1])   # 隐式索引获取元素
# 使用显示索引进行切片获取
print(f"使用显式索引获取元素：\n{s.loc['a':'c']}")
print(f"使用隐式索引获取元素：\n{s.iloc[1:3]}")
# 使用at[]和iat[]获取元素    无法进行切片
print(s.at['a'])
print(s.iat[0])
```

```python
# 访问数据
print(s['a'])
# 使用布尔索引访问数据
print(s[s<3])
s['f'] = 6
print(s.head())     # 取出数据的前5行
print(s.tail())     # 取出数据的后5行
```

```python
'''
案例：
创建一个包含10名学生数学成绩的Series，成绩范围在50-100之间。
计算平均分、最高分、最低分，并找出高于平均分的学生人数。
'''
# 生成随机成绩
np.random.seed(42)
scores = pd.Series(np.random.randint(50, 101, 10), index=["学生"+str(i) for i in range(1, 11)])
# 计算平均分/最高分/最低分
print(f"平均分：{scores.mean()}，最高分：{scores.max()}，最低分：{scores.min()}")
# 高于平均分的学生人数
mean = scores.mean()
# 这里可以使用count()或者len()或者直接.size属性
print(scores[scores > mean].count())
print(len(scores[scores > mean]))
print(scores[scores > mean].size)
```

```python
'''
案例2：
给定某城市一周每天的最高温度Series，完成以下任务：
找出温度超过30度的天数
计算平均温度
将温度从高到低排序
找出温度变化最大的两天
'''
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33],
index=['周一', '周二', '周三', '周四', '周五', '周六', '周日'])
# 找出温度超过30度的天数
print(f"温度超过30度的天数为：{temperatures[temperatures > 30].count()}")
# 计算平均温度
print(temperatures.mean())
# 将温度从高到低排序
print(temperatures.sort_values(ascending=False))
# 找出温度变化最大的两天
# temp_change = [abs(temperatures.iloc[i] - temperatures.iloc[i-1]) for i in range(1,len(temperatures))]  # 运用Python的生成器表达式
# print(temp_change)
t3 = abs(temperatures.diff())     # Series 自带方法diff()，计算series的变化值
print(f"温度变化最大的两天：{t3.sort_values(ascending=False).index[:2].tolist()}")
```

```python
'''
案例3：
给定某股票连续10个交易日的收盘价Series：
计算每日收益率（当日收盘价/前日收盘价 - 1）
找出收益率最高和最低的日期
计算波动率（收益率的标准差）
'''
prices = pd.Series([102.3, 103.5, 105.1, 104.8, 106.2, 107.0, 106.5, 108.1,
109.3, 110.2], index=pd.date_range('2023-01-01', periods=10))
print(prices)
# 计算每日收益率（当日收盘价/前日收盘价 - 1）
print(f"每日收益率：\n{prices.pct_change()}")
```

```python
# 收益率最高的日期
pct = prices.pct_change()
print(pct.idxmax())
# 收益率最低的日期
print(pct.idxmin())
# 波动率
print(pct.std())
```

```python
'''
某产品过去12个月的销售量Series：
计算季度平均销量（每3个月为一个季度）
找出销量最高的月份
计算月环比增长率
找出连续增长超过2个月的月份
'''
sales = pd.Series([120, 135, 145, 160, 155, 170, 180, 175, 190, 200, 210, 220],
index=pd.date_range('2022-01-01', periods=12, freq='MS'))
# 计算季度平均销量（每3个月为一个季度）
# resample：对时间索引的 Series/DataFrame，按指定时间频率重新分组并聚合数据。
quarter_avg = sales.resample("QE").mean()
print(quarter_avg)
```

```python
# 找出销量最高的月份
print(sales.idxmax())
# 计算月环比增长率 = (当月-上月)/上月
print(sales.pct_change())
# 找出连续增长超过2个月的月份
# 滑动窗口  稍微有点绕
grow = sales.pct_change()
increase = grow > 0
increase.rolling(3).sum()
increase[increase.rolling(3).sum() >= 3].index.tolist()
```

```python
'''
某商店每小时销售额 Series：
按天重采样计算每日总销售额
计算每天营业时间（8:00-22:00）和非营业时间的销售额比例
找出销售额最高的 3 个小时
'''
np.random.seed(42)
hourly_sales = pd.Series(np.random.randint(0, 100, 24),
index=pd.date_range('2025-01-01', periods=24, freq='h'))
# 按天重采样计算每日总销售额
daily_sales = hourly_sales.resample("D").sum()
print(daily_sales)
```

```python
# 计算每天营业时间（8:00-22:00）和非营业时间的销售额比例
business_sales = hourly_sales.between_time('8:00','22:00').sum()   # 筛选一段时间内的series    营业时间
non_business_sales = hourly_sales.between_time('23:00','7:00').sum()
print(f"营业时间销售额占比：{business_sales / non_business_sales}")
# 找出销售额最高的 3 个小时    使用nlargest()方法
print(hourly_sales.nlargest(3))
```
